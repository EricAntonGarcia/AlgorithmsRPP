```python
#Creating the cPOS5 reaction
atp_c = model.metabolites.get_by_id('atp_c')
nadh_c = model.metabolites.get_by_id('nadh_c')
adp_c = model.metabolites.get_by_id('adp_c')
nadph_c = model.metabolites.get_by_id('nadph_c')
h_c = model.metabolites.get_by_id('h_c')

cPOS5 = Reaction('cPOS5')
cPOS5.name = 'Cytosolic NADH Kinase from Saccharomyces cerevisiae'
cPOS5.add_metabolites({atp_c: -1, 
                           nadh_c: -1, 
                           adp_c: 1, 
                           nadph_c: 1,
                               h_c: 1})
cPOS5.bounds = (0,0)
print (cPOS5.reaction)
cPOS5

# from Saccharomyces. Found in Brenda ('for references in articles please use BRENDA:EC2.7.1.86')
# Also in Uniprot with ID Q06892 and ID YPL188W (Saccharomyces Genome Database)
# It is a kinase from the mitochondrial matrix (Outten and Culotta, 2003)
# Although it is a mitochondrial enzyme in Saccharomyces, Marius et al. (2) added a tag to express it in the cytosol
# you sure about the above? It says in page of his thesis that it was truncated: they removed the first 49 bp of the
# gene in order to remove the peptide leading sequence, ensuring that it would stay in the cytosol and not go to the
# mitochondria
```

```python
# Creating the cPntAB reaction

nad_c = model.metabolites.get_by_id('nad_c')
nadh_c = model.metabolites.get_by_id('nadh_c')
nadp_c = model.metabolites.get_by_id('nadp_c')
nadph_c = model.metabolites.get_by_id('nadph_c')

cPntAB = Reaction('cPntAB')
cPntAB.name = 'Cytosolic Nicotinamide Nucleotide Transhydrogenase'

cPntAB.add_metabolites({
    nadh_c: -1,
    nadp_c: -1,
    nad_c: 1,
    nadph_c: 1
})

cPntAB.bounds = (0, 0)

print(cPntAB.reaction)
cPntAB
```

```python
# Creating the cNoxE reaction

nad_c = model.metabolites.get_by_id('nad_c')
nadh_c = model.metabolites.get_by_id('nadh_c')
o2_c = model.metabolites.get_by_id('o2_c')
h2o_c = model.metabolites.get_by_id('h2o_c')
h_c = model.metabolites.get_by_id('h_c')

cNoxE = Reaction('cNoxE')
cNoxE.name = 'Water-forming NADH Oxidase from Lactococcus lactis'

cNoxE.add_metabolites({
    nadh_c: -1,
    h_c: -1,
    o2_c: -0.5,
    nad_c: 1,
    h2o_c: 1
})

cNoxE.bounds = (0, 0)

print(cNoxE.reaction)
cNoxE
```

```python
#Creating the HcAtOX reaction
q6h2_m = model.metabolites.get_by_id('q6h2_m')
o2_m = model.metabolites.get_by_id('o2_m')
q6_m = model.metabolites.get_by_id('q6_m')
h2o_m = model.metabolites.get_by_id('h2o_m')

AtOX = Reaction('AtOX')
AtOX.name = 'Alternative oxidase from Histoplasma capsulatum'
AtOX.add_metabolites({q6h2_m: -2,
                       o2_m: -1,
                       q6_m: 2,
                       h2o_m: 2})
AtOX.bounds = (0,0)
print(AtOX.reaction)

# UniProt ID: Q9Y711 (Johnson et al., 2003)
# Possible Brenda ID for reaction stoichiometry: EC 1.10.3.11
# It is an enzyme from the mitochondrial inner membrane
```

```python
print (len(model.reactions))
model.add_reactions([cNoxE])
model.add_reactions([cPOS5])
model.add_reactions([AtOX])
model.add_reactions([cPntAB])
print (len(model.reactions))
```

```python
pfba_WT = cobra.flux_analysis.pfba(model)
```

```python
pfba_WT.fluxes['Ex_biomass']
```

```python
WT_growth = pfba_WT.fluxes["Ex_biomass"]

biomass_values = np.linspace(
    0,
    WT_growth,
    50
)
```

```python
Growth = model.reactions.get_by_id("Ex_biomass")
ROL = model.reactions.get_by_id("Ex_rol")

conditions = {
    "WT": None,
    "NOXE": "cNoxE",
    "POS5": "cPOS5",
    "AtOX": "AtOX",
    "PntAB": "cPntAB"
}

intervention_flux = 1

# Save original bounds and objective
original_growth_bounds = Growth.bounds
original_rol_bounds = ROL.bounds
original_objective = model.objective
original_direction = model.objective_direction

original_candidate_bounds = {
    reaction_id: model.reactions.get_by_id(reaction_id).bounds
    for reaction_id in conditions.values()
    if reaction_id is not None
}

results = []

try:
    for condition, reaction_id in conditions.items():

        # Close all four added reactions first
        for candidate_id in original_candidate_bounds:
            model.reactions.get_by_id(candidate_id).bounds = (0, 0)

        # Activate only the reaction belonging to this condition
        if reaction_id is not None:
            model.reactions.get_by_id(reaction_id).bounds = (
                intervention_flux,
                intervention_flux
            )

        for biomass_value in biomass_values:

            Growth.bounds = (biomass_value, biomass_value)
            model.objective = ROL

            # Maximize ROL
            model.objective_direction = "max"
            sol_max = model.optimize()

            rol_max = (
                sol_max.fluxes["Ex_rol"]
                if sol_max.status == "optimal"
                else np.nan
            )

            # Minimize ROL
            model.objective_direction = "min"
            sol_min = model.optimize()

            rol_min = (
                sol_min.fluxes["Ex_rol"]
                if sol_min.status == "optimal"
                else np.nan
            )

            results.append({
                "Condition": condition,
                "Biomass": biomass_value,
                "ROL_max": rol_max,
                "ROL_min": rol_min,
                "Status_max": sol_max.status,
                "Status_min": sol_min.status
            })

finally:
    # Restore the model's original state
    Growth.bounds = original_growth_bounds
    ROL.bounds = original_rol_bounds
    model.objective = original_objective
    model.objective_direction = original_direction

    for reaction_id, bounds in original_candidate_bounds.items():
        model.reactions.get_by_id(reaction_id).bounds = bounds

product_envelopes_5conditions = pd.DataFrame(results)

print(product_envelopes_5conditions.head(20))
```

```python
print(product_envelopes_5conditions.head(100))
```

```python
conditions = ["WT", "NOXE", "POS5", "AtOX", "PntAB"]

plt.figure(figsize=(9, 5))

for condition in conditions:

    data = product_envelopes_5conditions[
        product_envelopes_5conditions["Condition"] == condition
    ].sort_values("Biomass")

    # Maximum ROL
    plt.plot(
        data["Biomass"],
        data["ROL_max"],
        label=f"{condition} max"
    )

    # Minimum ROL
    plt.plot(
        data["Biomass"],
        data["ROL_min"],
        linestyle="--",
        label=f"{condition} min"
    )

plt.xlabel("Biomass production")
plt.ylabel("ROL production")
plt.title("Product envelope: WT and engineered strains")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
for condition in ["NOXE", "POS5", "AtOX", "PntAB"]:

    plt.figure(figsize=(7, 5))

    for label in ["WT", condition]:

        data = product_envelopes_5conditions[
            product_envelopes_5conditions["Condition"] == label
        ].sort_values("Biomass")

        plt.plot(
            data["Biomass"],
            data["ROL_max"],
            label=f"{label} max"
        )

        plt.plot(
            data["Biomass"],
            data["ROL_min"],
            linestyle="--",
            label=f"{label} min"
        )

    plt.xlabel("Biomass production")
    plt.ylabel("ROL production")
    plt.title(f"Product envelope: WT vs {condition}")
    plt.legend()
    plt.tight_layout()
    plt.show()
```
